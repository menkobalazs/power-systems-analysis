## Kapacitással kapcsolatos paraméterek
- `p_nom` (float) – Nominal power for limits on p in optimization. Ignored if `p_nom_extendable`=`True`.
- Nominal power of the generator module (e.g. fixed unit size of a nuclear power plant). Introduces integer variables if set.
- `p_nom_extendable` (bool) – Switch to allow capacity p_nom to be extended in optimization.
- `p_nom_min`/`p_nom_max` - If p_nom is extendable in optimization, set its minimum/maximum value.
- `p_nom_set` - If p_nom is extendable in optimization, set its value.
- `p_min_pu`/`p_max_pu` -  The min/max output for each snapshot per unit of p_nom for the optimization (e.g. a minimal dispatch level for conventional power plants). Note that if committable=False and p_min_pu>0, this represents a must-run condition.
- `p_set` - Active power set point (for optimisation and power flow)
- `p_init` - Active power used as a starting point for limiting the ramp in the very first snapshot (for optimisation)
- `e_sum_min`/`e_sum_max` - The min/max total energy produced during a single optimization horizon.

## Pénzügyi paraméterek
- `marginal_cost(_quadratic)` - (Quadratic) marginal cost of production of 1 MWh.
- `capital_cost` - Fixed period costs of extending p_nom by 1 MW, Used directly in optimization unless `overnight_cost` is specified.
- `overnight_cost` – Overnight (upfront) investment cost per MW. If specified, PyPSA calculates annuity using `discount_rate` and `lifetime`. Takes precedence over capital_cost.
- `discount_rate` – Discount rate for annuity calculation when using overnight_cost. Supports 0% rate (simple depreciation).
- `fom_cost` – Fixed period operation and maintenance costs per MW, added to annuitized investment cost
- `efficiency` (Series) - Ratio output and primary energy carrier input
