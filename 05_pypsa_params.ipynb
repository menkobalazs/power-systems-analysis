# PyPSA - Network & Optimizations
## Generator and Storage Unit Parameters
---

# [Generator parameters](https://docs.pypsa.org/latest/api/components/types/generators/#pypsa.components.Generators.add)

## 1. Alapvető hálózati adatok

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `bus` | `str` | Annak a busznak a neve, amelyhez a generátor csatlakozik. |
| `control` | `str` | P, Q, V szabályozási stratégia a teljesítményáramláshoz ("PQ", "PV" vagy "Slack"). Csak AC és DC buszoknál releváns. |
| `carrier` | `str` | Elsődleges energiahordozó (pl. szén, gáz, szél, nap). |
| `sign` | `float` | Az irányultság jele (pozitív termelésnél, negatív fogyasztásnál). |
| `active` | `bool` | Meghatározza, hogy az erőmű szerepeljen-e az optimalizálásban. |
| `type` | `str` | Placeholder a generátor típusának (nincs implementálva). |

## 2. Kapacitás

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `p_nom` | `float` | Névleges teljesítmény. Figyelmen kívül marad, ha a `p_nom_extendable=True`. |
| `p_nom_extendable` | `bool` | Kapcsoló, amely lehetővé teszi a kapacitás (p_nom) bővítését az optimalizálás során. |
| `p_nom_min` / `_max`  | `float` or `Series` | A névleges teljesítmény alsó/felső korlátja. |
| `p_nom_set` | `float` or `Series` | A névleges teljesítmény fixen beállított értéke, ha a kapacitás bővíthető. |
| `p_nom_mod` | `float` | A generátormodul névleges teljesítménye (pl. egy atomerőmű fix blokkmérete). Beállítása egész számú (integer) változókat vezet be. |
| `efficiency` | `float` or `Series` | Hatásfok. |

## 3. Üzemeltetés és korlátok

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `p_set` / `q_set` | `float` or `Series` | Aktív és meddő teljesítmény (fix érték a számításhoz). |
| `p_min_pu` / `p_max_pu` | `float` | Minimális/maximális kimenet snapshot-onként, a névleges teljesítmény arányában (per-unit). |
| `e_sum_min` / `e_sum_max` | `float` | A teljes optimalizálási időszak alatt termelhető minimális/maximális összenergia. |
| `ramp_limit_up` / `_down` | `float` or `Series` | Maximális teljesítmény-növekedés/csökkenés két snapshot között (per-unit). |
| `p_init` | `float` | Kezdő teljesítmény az első snapshot előtt a meredekség korlátozásához. |
| `weight` | `float` | Generátor súlyozása. Csak abban az esetben használható, ha a hálózat klaszterezett. |

## 4. Költségek

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `marginal_cost` | `float` or `Series` | Határköltség (1 MWh termelésének költsége). |
| `marginal_cost_quadratic` | `float` or `Series` | Kvadratikus (másodfokú) határköltség. |
| `capital_cost` | `float` | Fajlagos beruházási költség (annuitás formájában, pl. EUR/MW/év). |
| `overnight_cost` | `float` | Azonnali (upfront) beruházási költség MW-onként. |
| `lifetime` | `float` | A generátor élettartama években. |
| `discount_rate` | `float` | Discount rate az annuitás számításához. |
| `fom_cost` | `float` | Fix üzemeltetési és karbantartási (fixed period operation and maintenance) költség (MW-onként). |
| `build_year` | `int` | A generátor építésének éve. |

## 5. Unit Commitment

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `committable` | `bool` | Egységelkötelezési korlátok alkalmazása (ki/be kapcsolhatóság). |
| `start_up_cost` / `shut_down_cost` | `float` | Indítási és leállítási költségek. |
| `stand_by_cost` | `float` | Készenléti költség (akkor is merül fel, ha a gép be van kapcsolva, de nem termel). |
| `min_up_time` / `min_down_time` | `int` | Minimális bekapcsolva/kikapcsolva tartási idő (snapshot-ok száma). |
| `up_time_before` / `down_time_before` | `int` | Hány snapshot-on át volt be/ki kapcsolva a kezdőpont előtt. |
| `ramp_limit_start_up` / `_shut_down` | `float` | Maximális meredekség indításkor és leállításkor (per-unit). |

## 6. Kimenetek

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `p` / `q` | `Series` | Kiszámított aktív és meddő teljesítmény snapshot-ként. |
| `p_nom_opt` | `float` | Az optimalizált (kiszámított ideális) névleges kapacitás. |
| `status` | `Series` | Üzemállapot (1 - be, 0 - ki) – csak ha `committable=True`. |
| `start_up` / `shut_down` | `bool` / `int` | Indítási/leállítási esemény az adott snapshot-ban. |
| `mu_p_set ` | `Series` | Shadow price of fixed power generation p_set. |
| `mu_upper` / `_lower` | `float` | A névleges kapacitás korlátainak árnyékárai (shadow price). |
| `mu_ramp_limit_up` / `_down` | `float` | A meredekségi korlátok árnyékárai. |

# [Storage Unit paramaters](https://docs.pypsa.org/latest/api/components/types/storage-units/#pypsa.components.StorageUnits.add)


## 1. Alapvető hálózati adatok

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `bus` | `str` | Annak a busznak a neve, amelyhez a tároló csatlakozik. |
| `control` | `str` | P, Q, V szabályozási stratégia a teljesítményáramláshoz ("PQ", "PV" vagy "Slack"). |
| `carrier` | `str` | Elsődleges energiahordozó (pl. szivattyús tározó, lítium-ion, hidrogén). |
| `sign` | `float` | Az irányultság jele (pozitív termelésnél, negatív fogyasztásnál). |
| `active` | `bool` | Meghatározza, hogy az elem szerepeljen-e az optimalizálásban. |
| `type` | `str` | Placeholder a tároló típusának (nincs implementálva). |

## 2. Kapacitás

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `p_nom` | `float` | Névleges teljesítmény. Figyelmen kívül marad, ha a `p_nom_extendable=True`. |
| `p_nom_extendable` | `bool` | Kapcsoló, amely lehetővé teszi a kapacitás (`p_nom`) bővítését az optimalizálás során. |
| `p_nom_min` / `_max`  | `float` | A névleges teljesítmény alsó/felső korlátja optimalizálásnál. |
| `p_nom_set` | `float` | A névleges teljesítmény fix értéke bővíthető kapacitás esetén. |
| `p_nom_mod` | `float` | A modul névleges teljesítménye (egész számú változókat vezet be). |
| `max_hours` | `float` | Maximális töltöttségi kapacitás órákban kifejezve a névleges teljesítmény (`p_nom`) arányában. |

## 3. Tárolási karakterisztika és működés

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `p_set` / `q_set` | `float / Series` | Aktív és meddő teljesítmény alapjel (fix érték). LOPF esetén a nettó teljesítményt korlátozza (`p_dispatch` - `p_store`). |
| `p_min_pu` / `p_max_pu` | `float / Series` | Minimális/maximális kimenet a névleges teljesítmény arányában. Negatív érték = betárolás, pozitív = kisütés. |
| `p_dispatch_set` | `float / Series` | Aktív teljesítmény-leadási (kisütési) alapjel (csak optimalizáláshoz). |
| `p_store_set` | `float / Series` | Aktív teljesítmény-betárolási (töltési) alapjel (csak optimalizáláshoz). |
| `efficiency_store` | `float / Series` | A betárolás (töltés) hatásfoka. |
| `efficiency_dispatch` | `float / Series` | A leadás (kisütés) hatásfoka. |
| `standing_loss` | `float / Series` | Óránkénti töltöttségi veszteség (önkisülés). |
| `inflow` | `float / Series` | Külső betáplálás a tárolóba (pl. folyóvíz hozama víztározónál). |

## 4. Töltöttségi állapot

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `state_of_charge_initial` | `float` | Töltöttségi állapot az optimalizálás kezdete előtt. |
| `state_of_charge_initial_per_period` | `bool` | Ha True, a kezdő töltöttség minden beruházási időszak elején érvényesül. |
| `state_of_charge_set` | `float / Series` | Előre meghatározott töltöttségi állapot alapjelek a snapshotokhoz. |
| `cyclic_state_of_charge` | `bool` | Ciklikus kényszer: a záró töltöttségnek meg kell egyeznie a kezdővel (soc[-1] = soc[0]). |
| `cyclic_state_of_charge_per_period` | `bool` | A ciklikus kényszer minden beruházási időszakra külön-külön vonatkozik. |

## 5. Költségek

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `marginal_cost` | `float / Series` | Kisütés határköltsége (1 MWh termelésének költsége). |
| `marginal_cost_quadratic` | `float / Series` | Kvadratikus határköltség a kisütésnél. |
| `marginal_cost_storage` | `float / Series` | 1 MWh energia tárolásának határköltsége egy órán keresztül. |
| `spill_cost` | `float / Series` | A túlfolyás (spillage) költsége 1 MWh-ra vetítve. |
| `capital_cost` | `float` | Fajlagos beruházási költség (annuitás, pl. EUR/MW/év). |
| `overnight_cost` | `float` | Azonnali beruházási költség MW-onként. |
| `lifetime` | `float` | A tároló élettartama években. |
| `discount_rate` | `float` | Leszámítolási ráta az annuitás számításához. |
| `fom_cost` | `float` | Fix üzemeltetési és karbantartási költség (MW-onként). |
| `build_year` | `int` | Építési év. |

## 6. Kimenetek

| Paraméter | Típus | Leírás |
| :--- | :--- | :--- |
| `p` / `q` | `Series` | Kiszámított nettó aktív és meddő teljesítmény (pozitív, ha nettó generálás). |
| `p_dispatch` | `Series` | Kiszámított aktív teljesítmény-leadás (kisütés). |
| `p_store` | `Series` | Kiszámított aktív teljesítmény-felvétel (töltés). |
| `state_of_charge` | `Series` | Az optimalizált töltöttségi állapot snapshot-onként. |
| `spill` | `Series` | Kiszámított túlfolyás/energiakiengedés értéke. |
| `p_nom_opt` | `float` | Az optimalizált névleges kapacitás. |
| `mu_upper` / `_lower` | `float` | A névleges teljesítménykorlátok árnyékárai. |
| `mu_state_of_charge_set` | `Series` | A rögzített töltöttségi állapot (`state_of_charge_set`) árnyékára. |
| `mu_energy_balance` | `Series` | A tárolási energiamérleg-egyenletek árnyékára (a tárolt energia értéke). |
