# Data Comparison

```python
import pandas as pd
from utils import data_path

import pypsa

import warnings
warnings.filterwarnings("ignore", category=UserWarning, module="openpyxl")
```

## Potentials

```python
potentials_generator = pd.read_excel(data_path, sheet_name='potentials_generator', index_col=0, na_values='None')
potentials_generator = potentials_generator.loc['p_nom_max'] 
technologies_generator = potentials_generator.keys().to_numpy()

entsoe_pots = pd.read_excel("data/entsoe/net_generation_capacity_2025.xlsx")
entsoe_pots = entsoe_pots[entsoe_pots['Country']=='HU'][['Category', 'ProvidedValue']]
entsoe_pots = entsoe_pots.set_index(keys=entsoe_pots['Category']).drop(columns=['Category'])
entsoe_pots = pd.Series(entsoe_pots.to_dict()['ProvidedValue'])

entsoe_pots = entsoe_pots.rename({'Fossil Brown coal/Lignite': 'Fossil Lignite'})
entsoe_pots = entsoe_pots.rename({'Hydro Run-of-river and poundage': 'Hydro Run-of-river'})

pd.DataFrame({'Our Dataset (~2050)': potentials_generator, 'ENTSO-e (2025)': entsoe_pots, 'Difference': potentials_generator-entsoe_pots})
```
